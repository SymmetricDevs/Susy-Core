```python
from pathlib import Path

java_template = """package supersymmetry.common.blocks.bmrf;

import net.minecraft.block.SoundType;
import net.minecraft.block.material.Material;
import net.minecraft.block.state.IBlockState;
import net.minecraft.item.ItemStack;
import net.minecraft.util.IStringSerializable;

import org.jetbrains.annotations.NotNull;

import gregtech.api.block.IStateHarvestLevel;
import gregtech.api.block.VariantBlock;

public class BlocksS{group}BMRF extends VariantBlock<BlocksS{group}BMRF.S{group}BMRFBlockType> {{

    public BlocksS{group}BMRF() {{
        super(Material.ROCK);
        this.setHardness(3.0F);
        this.setResistance(5.0F);
        this.setSoundType(SoundType.STONE);
        this.setTranslationKey("s{group}bmrf_blocks");
    }}

    @Override
    public IBlockState getStateFromMeta(int meta) {{
        return this.getDefaultState().withProperty(VARIANT, VALUES[meta % VALUES.length]);
    }}

    @Override
    public int getMetaFromState(@NotNull IBlockState state) {{
        return state.getValue(VARIANT).ordinal();
    }}

    @Override
    public ItemStack getItemVariant(S{group}BMRFBlockType variant, int amount) {{
        return new ItemStack(this, amount, variant.ordinal());
    }}

    @Override
    public int damageDropped(@NotNull IBlockState state) {{
        return state.getValue(VARIANT).ordinal();
    }}

    public static enum S{group}BMRFBlockType implements IStringSerializable, IStateHarvestLevel {{

{enum_values};

        private final String name;
        private final int harvestLevel;

        private S{group}BMRFBlockType(String name, int harvestLevel) {{
            this.name = name;
            this.harvestLevel = harvestLevel;
        }}

        @Override
        public int getHarvestLevel(IBlockState iBlockState) {{
            return this.harvestLevel;
        }}

        @Override
        public String getName() {{
            return this.name;
        }}
    }}
}}
"""

# Generates BlocksS16BMRF.java through BlocksS24BMRF.java.
# The final class has 10 variants because textures stop at bmrf432.
output_dir = Path(".")
for group in range(16, 25):
    count = min(16, 433 - (295 + (group - 16) * 16))
    enum_values = ",\n".join(
        f'        S{group}BMRF{i}("s{group}bmrf{i}", 2)'
        for i in range(1, count + 1)
    )
    path = output_dir / f"BlocksS{group}BMRF.java"
    path.write_text(
        java_template.format(group=group, enum_values=enum_values),
        encoding="utf-8"
    )
    print(f"Created {path.name} ({count} variants)")
```
